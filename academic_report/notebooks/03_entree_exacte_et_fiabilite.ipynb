# Entrée exacte et fiabilité

Ce notebook vérifie les modèles temps-vers-entrée et la fiabilité du signal court horizon.

```
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
while not (ROOT / 'runs').exists():
    ROOT = ROOT.parent
pd.set_option('display.max_columns', 50)
ROOT

causal = pd.read_csv(ROOT / 'runs/exp_105_exact_entry_forecast_rerun/metrics/exact_entry_causal_summary.csv')
cols = ['objective', 'base_model_name', 'score_col', 'threshold', 'persistence_windows', 'pre_entry_recall_mean', 'early_03_recall_mean', 'early_05_recall_mean', 'event_precision_mean', 'false_alarms_per_min_mean']
causal[causal['split'].eq('test')][cols].sort_values('early_05_recall_mean', ascending=False).head(12)
```

```
rel = pd.read_csv(ROOT / 'runs/exp_105_exact_entry_forecast_rerun/metrics/exact_entry_reliability_summary.csv')
rel_cols = ['base_model_name', 'score_col', 'threshold', 'n_windows_total', 'entry_le_02_rate_mean', 'entry_le_03_rate_mean', 'entry_le_05_rate_mean', 'mean_time_to_entry_s_mean']
rel[(rel['split'].eq('test')) & (rel['threshold'].ge(0.9))][rel_cols].sort_values('entry_le_03_rate_mean', ascending=False).head(12)
```

```
best_stop = rel[(rel['split'].eq('test')) & (rel['base_model_name'].eq('survival_gru_exact_entry')) & (rel['score_col'].eq('score_by_02')) & (rel['threshold'].eq(0.95))]
best_stop[rel_cols]
```
